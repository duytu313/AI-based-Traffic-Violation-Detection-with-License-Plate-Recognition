```python
# 0. Cài đặt thư viện ultralytics nếu Colab chưa có


import torch
import torchvision.transforms as transforms
from torchvision.datasets import ImageFolder
from torch.utils.data import DataLoader
from sklearn.metrics import classification_report, confusion_matrix
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
from ultralytics import YOLO

# 1. Cấu hình thiết bị (GPU nếu có, không thì dùng CPU)
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Đang sử dụng thiết bị: {device}")

# 2. Định nghĩa các bước tiền xử lý ảnh
transform_test = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
])

# 3. Tải dữ liệu tập Validation
test_dir = '/content/Color_Vehicle/valid'

try:
    test_dataset = ImageFolder(root=test_dir, transform=transform_test)
    test_loader = DataLoader(test_dataset, batch_size=1, shuffle=False, num_workers=2)

    # Định nghĩa danh sách class theo đúng thứ tự bạn cung cấp
    class_names = ['beige', 'black', 'blue', 'brown', 'gold', 'green', 'grey', 'orange', 'pink', 'purple','red', 'silver', 'tan', 'white', 'yellow']

    # Kiểm tra lại xem số lượng thư mục thực tế có khớp không
    if len(test_dataset.classes) != len(class_names):
        print(f"⚠️ Cảnh báo: Số lượng thư mục thực tế ({len(test_dataset.classes)}) khác với danh sách bạn nhập ({len(class_names)}). Sẽ tự động dùng danh sách gốc từ thư mục.")
        class_names = test_dataset.classes

    print(f"Tìm thấy {len(test_dataset)} ảnh test thuộc {len(class_names)} lớp.")
except Exception as e:
    print(f"Lỗi load dữ liệu: {e}. Hãy kiểm tra lại đường dẫn 'test_dir'.")

# 4. Khởi tạo và Load Model YOLOv8
path_to_model = '/content/vehicle_color_n_cls.pt'
try:
    model = YOLO(path_to_model)
    print("Load model YOLOv8 thành công!")
except Exception as e:
    print(f"Lỗi load model: {e}")

# 5. Tiến hành dự đoán trên tập test
all_preds = []
all_labels = []

print("Đang chạy dự đoán trên toàn bộ tập dữ liệu...")
for images, labels in test_loader:
    img_tensor = images[0]
    img_pil = transforms.ToPILImage()(img_tensor)

    results = model(img_pil, verbose=False)
    for r in results:
        pred_class_idx = r.probs.top1
        all_preds.append(pred_class_idx)
        all_labels.append(labels.item())

# 6. Xuất báo cáo chi tiết (Accuracy, Precision, Recall, F1-score)
print("\n" + "="*20 + " BÁO CÁO ĐÁNH GIÁ (CLASSIFICATION REPORT) " + "="*20)
print(classification_report(all_labels, all_preds, target_names=class_names))

# 7. Vẽ Ma trận nhầm lẫn (Confusion Matrix) kích thước lớn cho 14 màu
cm = confusion_matrix(all_labels, all_preds)
plt.figure(figsize=(14, 12))  # Tăng kích thước hình để không bị lấp chữ
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=class_names, yticklabels=class_names)
plt.xlabel('(Predicted)', fontsize=12, fontweight='bold')
plt.ylabel('(Actual)', fontsize=12, fontweight='bold')
plt.title('Confusion Matrix - Color Vehicle Classification (15 Colors)', fontsize=14, fontweight='bold')
plt.xticks(rotation=45, ha='right')
plt.yticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import cv2
import torch
import numpy as np
from pathlib import Path
from ultralytics import YOLO

# --- CẤU HÌNH ---

# 1. Đường dẫn đến các mô hình
path_to_det_model = 'yolov8n.pt' # Dùng model YOLOv8-Detection chuẩn để nhận dạng xe
path_to_cls_model = '/content/vehicle_color_n_cls.pt' # Model phân loại màu sắc của bạn

# 2. Ngưỡng tự tin cho nhận dạng xe (Detection)
det_conf = 0.40
# 3. Định nghĩa các class xe mà model Detection hỗ trợ (theo chuẩn COCO dataset)
# COCO class IDs: 2=car, 3=motorcycle, 5=bus, 7=truck.
VEHICLE_CLASSES = [2, 3, 5, 7]

# 4. Ngưỡng tự tin cho phân loại màu (Classification) - không dùng trực tiếp trong code nhưng r.conf sẽ thể hiện
cls_conf_threshold = 0.30

# 5. Ảnh test
image_path = '/content/bien_so.jpg'

# --- KHỞI TẠO CÁC MODEL ---
print("Đang load các mô hình...")
# Dùng GPU nếu có, không thì dùng CPU
device = 'cuda' if torch.cuda.is_available() else 'cpu'

try:
    det_model = YOLO(path_to_det_model).to(device)
    print(f"Load mô hình Detection '{path_to_det_model}' thành công trên '{device}'!")

    cls_model = YOLO(path_to_cls_model).to(device)
    print(f"Load mô hình Classification '{path_to_cls_model}' thành công trên '{device}'!")
    print("-" * 50)
except Exception as e:
    print(f"Lỗi load mô hình: {e}")
    exit()

# Định nghĩa các class màu chuẩn từ model classification của bạn (để lấy r.names)
color_names = cls_model.names

# --- TIẾN HÀNH XỬ LÝ ẢNH ---

# Đọc ảnh bằng OpenCV để vẽ hiển thị lên màn hình
img_orig = cv2.imread(image_path)
if img_orig is None:
    print(f"Lỗi: Không tìm thấy ảnh tại đường dẫn '{image_path}'. Vui lòng kiểm tra lại.")
    exit()

# Tạo một bản sao để vẽ kết quả
img_result = img_orig.copy()

print(f"Đang xử lý ảnh: {Path(image_path).name}...")

# Bước 1: Dùng Model #1 (Detection) để tìm vị trí tất cả các xe
print("Bước 1: Chạy mô hình Nhận dạng vật thể (Detection)...")
det_results = det_model(img_orig, conf=det_conf, classes=VEHICLE_CLASSES, verbose=False) # Chỉ nhận dạng car, motorcycle, bus, truck

# Lấy dữ liệu các khung bao (bounding boxes)
boxes = det_results[0].boxes

vehicle_count = len(boxes)
print(f"Tìm thấy {vehicle_count} phương tiện trong ảnh.")
print("-" * 50)

if vehicle_count > 0:
    print("Bước 2: Cắt ảnh xe, chạy mô hình Phân loại màu sắc (Classification) và vẽ kết quả...")

    # Bước 2 & 3: Lặp qua từng chiếc xe được nhận dạng
    for i, box in enumerate(boxes):
        # Lấy tọa độ khung bao (xyxy) - ở định dạng integer
        x1, y1, x2, y2 = box.xyxy[0].cpu().numpy().astype(int)

        # Đảm bảo tọa độ nằm trong phạm vi ảnh (ngăn lỗi crop)
        x1 = max(0, x1)
        y1 = max(0, y1)
        x2 = min(img_orig.shape[1], x2)
        y2 = min(img_orig.shape[0], y2)

        # Thêm một chút biên để ảnh cắt rõ hơn (tùy chọn)
        crop_pad = 2 # pixels
        c_x1 = max(0, x1 - crop_pad)
        c_y1 = max(0, y1 - crop_pad)
        c_x2 = min(img_orig.shape[1], x2 + crop_pad)
        c_y2 = min(img_orig.shape[0], y2 + crop_pad)

        # Cắt ảnh chiếc xe
        vehicle_crop = img_orig[c_y1:c_y2, c_x1:c_x2]

        # Kiểm tra xem ảnh cắt có dữ liệu không (ngăn lỗi model classification)
        if vehicle_crop.size == 0:
            print(f"Xe #{i+1}: Ảnh cắt không có dữ liệu, bỏ qua.")
            continue

        # Dùng Model #2 (Classification) để dự đoán màu cho ảnh xe đã cắt
        cls_results = cls_model(vehicle_crop, verbose=False) # Không hiện log verbose

        # Lấy kết quả phân loại
        # Top1
        top1_idx = cls_results[0].probs.top1
        top1_name = color_names[top1_idx]
        top1_conf = cls_results[0].probs.top1conf.item() * 100

        print(f"Xe #{i+1}: {top1_name} ({top1_conf:.2f}%)")

        # Bước 3: Vẽ kết quả lên ảnh sao
        # 3.1: Vẽ khung bao (Bounding Box)
        cv2.rectangle(img_result, (x1, y1), (x2, y2), (0, 255, 0), 2) # Khung xanh lá, độ dày 2

        # 3.2: Chuẩn bị nội dung chữ (Tên màu + %)
        label_text = f"{top1_name} {top1_conf:.1f}%"

        # 3.3: Tính toán kích thước chữ để vẽ nền
        font_face = cv2.FONT_HERSHEY_SIMPLEX
        font_scale = 0.7 # Kích thước chữ
        font_thickness = 1
        label_size, base_line = cv2.getTextSize(label_text, font_face, font_scale, font_thickness)

        # 3.4: Đảm bảo nền chữ không bị tràn lên trên ảnh
        label_y1 = max(y1, label_size[1] + 10) # 10 là khoảng padding

       # 3.5: Vẽ khung nền cho chữ (Đổi thành màu TRẮNG: (255, 255, 255))
        cv2.rectangle(img_result,
                      (x1, label_y1 - label_size[1] - 10),
                      (x1 + label_size[0] + 10, label_y1 + base_line - 10),
                      (255, 255, 255),
                      cv2.FILLED)

        # 3.6: Vẽ chữ (Đổi thành màu ĐEN: (0, 0, 0))
        cv2.putText(img_result,
                    label_text,
                    (x1 + 5, label_y1 - 10),
                    font_face,
                    font_scale,
                    (0, 0, 0),
                    font_thickness,
                    cv2.LINE_AA)

else:
    print("Không tìm thấy bất kỳ phương tiện nào với ngưỡng tự tin hiện tại.")

# --- HIỂN THỊ KẾT QUẢ CUỐI CÙNG ---

# Chuyển hệ màu để matplotlib hiển thị đúng (OpenCV dùng BGR)
img_result_rgb = cv2.cvtColor(img_result, cv2.COLOR_BGR2RGB)

# Hiển thị ảnh
plt.figure(figsize=(10, 10))
plt.imshow(img_result_rgb)
plt.title(f"Simultaneous Vehicle Color Classification Results: Found {vehicle_count} vehicles", fontsize=14, fontweight='bold')
plt.axis('off') # Tắt trục tọa độ
plt.show()

# In thông tin tổng hợp ra console
print("-" * 50)
print(f"Hoàn tất xử lý ảnh: {Path(image_path).name}. Kết quả đã được vẽ lên ảnh hiển thị.")
```
